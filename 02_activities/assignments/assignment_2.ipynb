**Description**  
Given the `root` of a binary tree containing non-negative values, check whether it contains a duplicate value. If a duplicate exists, return the duplicate value. If there are multiple duplicates, return the one whose duplicate is closest to the root. If no duplicate exists, return -1.

**Examples**

* Example 3:

    ![](./images/q1_ex3.png)

    Input: `root = [10, 9, 8, 7]`

    Output: -1

**Problem 1 Starter Code**

You may assume that `root` is a binary tree that has been built, i.e., you can access a node/leaf by calling `root.left.right.left.right...`, so you don't have to provide a tree-building function to convert an input list to a binary tree as an object of the `TreeNode` class.

```python
from collections import deque

def is_duplicate(root) -> int:
  if root is None:
    return -1

  seen = set()
  queue = deque([root])

  while queue:
    node = queue.popleft()

    if node.val in seen:
      return node.val

    seen.add(node.val)

    if node.left:
      queue.append(node.left)

    if node.right:
      queue.append(node.right)

  return -1
```

```python
#for the input [0, 7, 2, 0, 5], the function processes the list recursively until it reaches the end. 
# As the recursive calls return, each number is checked.
# The non zero values 5, 2, and 7 are placed at the beginning of the result in a way that reconstructs their original order as [7, 2, 5]. 
# The two zeros are appended to the end. Therefore, the result is [7, 2, 5, 0, 0]
```


4. Copy the solution your partner wrote. 


```python
from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:
    # TODO
    result = []
    zero_count = 0

    # Go through each number in the list
    for num in nums:
        if num == 0:
            zero_count += 1
        else:
            result.append(num)

    # Add all zeros at the end
    for i in range(zero_count):
        result.append(0)

    return result


# Example 1
nums = [0, 1, 0, 3, 12]
print(move_zeros_to_end(nums))

# Example 2
nums = [4, 0, 5, 0, 0, 6]
print(move_zeros_to_end(nums))
```
