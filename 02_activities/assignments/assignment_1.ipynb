3. Based on the provided starter code, code the solution to your assigned problem here in Python. Note: each problem can be solved more simply if you use an abstract data type that is suitable for that problem. Try to find the solution with best time and space complexity.

````python
from typing import List

def first_duplicate(nums: List[int]) -> int:
    seen = set()

    for num in nums:
        if num in seen:
            return num
        seen.add(num)

    return -1
````


4. Walk through your Python code in plain words and explain why your solution works.


````python
# I created an empty set called seen to keep track of the numbers we have encountered so far. 
# The function looped through each number in the input list nums. If the current number is already in the seen set, it means it has found a duplicate, and return that number.
#  If not, the number is added to the seen set. If loop is completed without finding any duplicates, it return -1. 
# This worked because sets in Python have an average time complexity of O(1) for membership checks and insertions, making the overall time complexity of this function O(n), 
# where n is the length of the input list.
#  The space complexity is also O(n) in the worst case, when all numbers are unique and need to be stored in the set.
````


5. Explain the time and space complexity of your solution.


````python
# Time complexity: O(n) on average, because we go through the list only once. Checking and adding an element to a set normally takes O(1) time.
# Space complexity: O(n) because, in the worst case, we may need to store all the numbers in the set when no duplicates exist
````
